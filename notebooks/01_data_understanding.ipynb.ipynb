```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("synthetic_fraud_dataset.csv")
df.head(5)
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
print("Dataset shape:", df.shape)
```

```python
print("Column Names:")
for i, column in enumerate(df.columns, start=1):
    print(f"{i}. {column}")
```

```python
Unique_values = pd.DataFrame({
    "Column": df.columns,
    "Unique_Values": df.nunique(),
    "Unique_Percentage": (df.nunique() / len(df) * 100).round(2)
})

Unique_values
```

```python
categorical_columns = [
    "Transaction_Type",
    "Device_Type",
    "Location",
    "Merchant_Category",
    "Card_Type"
]

for column in categorical_columns:
    print("\n" + "=" * 60)
    print(f"{column}")
    print("=" * 60)
    print(df[column].unique())
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

print("Numerical columns:")
print(list(numeric_columns))
```

```python
numerical_summary = pd.DataFrame({
    "Mean": df[numeric_columns].mean(),
    "Median": df[numeric_columns].median(),
    "Std_Dev": df[numeric_columns].std(),
    "Minimum": df[numeric_columns].min(),
    "Q1": df[numeric_columns].quantile(0.25),
    "Q3": df[numeric_columns].quantile(0.75),
    "Maximum": df[numeric_columns].max(),
    "Skewness": df[numeric_columns].skew()
})

numerical_summary.round(2)
```

```python
for column in categorical_columns:
    print("\n" + "=" * 60)
    print(column)
    print("=" * 60)
    print(df[column].value_counts())
```

```python
print("\n" + "=" * 60)
print("Fraud Count")
print("=" * 60)
fraud_counts = df["Fraud_Label"].value_counts()

print(fraud_counts)
print("\n" + "=" * 60)
print("Fraud Percentage")
print("=" * 60)
fraud_percentage = df["Fraud_Label"].value_counts(normalize=True) * 100

print(fraud_percentage.round(2))
```

```python
print("=" * 70)
print("STAGE 1 - DATASET UNDERSTANDING SUMMARY")
print("=" * 70)

print(f"Rows: {df.shape[0]:,}")
print(f"Columns: {df.shape[1]}")
print(f"Missing Values: {df.isnull().sum().sum():,}")
print(f"Duplicate Rows: {df.duplicated().sum():,}")
print(f"Unique Transactions: {df['Transaction_ID'].nunique():,}")
print(f"Unique Users: {df['User_ID'].nunique():,}")
print(f"Fraud Transactions: {df['Fraud_Label'].sum():,}")
print(
    f"Fraud Rate: {df['Fraud_Label'].mean() * 100:.2f}%"
)

print("=" * 70)
```
